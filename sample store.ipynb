```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import dash as dash 

df=pd.read_csv("SampleSuperstore.csv")
print(df)

print(df.info())        # Check data types and missing values
print(df.describe())    # Summary statistics
print(df.isnull().sum())# Count missing values
print(df.duplicated().sum())  # Count duplicate values
print(df.columns)
print(df['State'].nunique())

df['Cost']=df['Sales']-df['Profit']
print(df.head())

state_cost=df.groupby('State')['Cost'].sum()

state_cost.plot(
      kind='bar',
    figsize=(17,8),
    color='aqua'
)
plt.title('Profit from State')
plt.xlabel('State')
plt.ylabel('Cost')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()




state_sales=df.groupby('State')['Sales'].sum().sort_values(ascending=False)

state_sales.plot(
    kind='bar',
    figsize=(15,10),
    color='red' 
)

plt.title('Sales by State')
plt.xlabel('State')
plt.ylabel('Sales')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

state_quantity=df.groupby('State')['Quantity'].sum().sort_values(ascending=False)

state_quantity.plot(
    kind='bar',
    figsize=(16,8),
    color='green'
)

plt.title('quantity imported by State')
plt.xlabel('State')
plt.ylabel('Quantity')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()

state_profit=df.groupby('State')['Profit'].sum().sort_values(ascending=False)

state_profit.plot(
    kind='bar',
    figsize=(17,8),
    color='blue'
)
plt.title('Profit from State')
plt.xlabel('State')
plt.ylabel('Profit')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()

category_profit=df.groupby('Category')['Profit'].sum()

category_profit.plot(
    kind='pie',
    autopct='%1.1f%%',   # Show percentages
    startangle=90,       # Rotate chart so it starts at the top
    shadow=True
)
plt.title('Profit by Category')
plt.ylabel(' ')
plt.show()

category_sales=df.groupby('Category')['Sales'].sum()

category_sales.plot(
    kind='pie',
    autopct='%1.1f%%',   # Show percentages
    startangle=90,       # Rotate chart so it starts at the top
    
)

plt.title('Sales by category')
plt.ylabel(' ')
plt.show()
```
