```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import dash as dash 

df=pd.read_csv("SampleSuperstore.csv")
print(df)

print(df.info())        # Check data types and missing values
print(df.describe())    # Summary statistics
print(df.isnull().sum())# Count missing values
print(df.duplicated().sum())  # Count duplicate values
print(df.columns)
print(df['State'].nunique())
state_sales=df.groupby('State')['Sales'].sum().sort_values(ascending=False)

state_sales.plot(
    kind='bar',
    figsize=(15,10),
    color='red' 
)

plt.title('Sales by State')
plt.xlabel('State')
plt.ylabel('Sales')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

state_quantity=df.groupby('State')['Quantity'].sum().sort_values(ascending=False)

state_quantity.plot(
    kind='bar',
    figsize=(16,8),
    color='green'
)

plt.title('quantity imported by State')
plt.xlabel('State')
plt.ylabel('Quantity')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()

state_profit=df.groupby('State')['Profit'].sum().sort_values(ascending=False)

state_profit.plot(
    kind='bar',
    figsize=(17,8),
    color='blue'
)
plt.title('Profit generated by State')
plt.xlabel('State')
plt.ylabel('Profit')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```
